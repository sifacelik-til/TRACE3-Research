```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set style for plots
plt.style.use("seaborn-v0_8")
sns.set_palette("husl")
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)

DATA_PROCESSED = Path("C:/Users/scelik/Desktop/TRACE3Code/data/raw/LSEG")
OUTPUT_PLOTS = Path("C:/Users/scelik/Desktop/TRACE3Code/plots")
OUTPUT_PLOTS.mkdir(parents=True, exist_ok=True)

# Load the LSEG ESG dataset (from your script's output)
# Assuming the file is named lseg_active_universe_10y.csv
try:
    lseg_df = pd.read_csv(DATA_PROCESSED / "lseg_active_universe_10y.csv", low_memory=False)
    print(f"📌 Dataset loaded: {lseg_df.shape[0]:,} rows × {lseg_df.shape[1]} columns")
except FileNotFoundError:
    # Try alternative filenames
    files = list(DATA_PROCESSED.glob("*.csv"))
    if files:
        lseg_df = pd.read_csv(files[0], low_memory=False)
        print(f"📌 Dataset loaded: {files[0].name} ({lseg_df.shape[0]:,} rows × {lseg_df.shape[1]} columns)")
    else:
        raise FileNotFoundError("No CSV files found in the outputs directory.")

# Display first few rows and column info
print("\n📌 **First 3 Rows:**")
display(lseg_df.head(3))
print("\n📌 **Column Data Types:**")
display(lseg_df.dtypes)

# %% [markdown]

"""
# 🌍 LSEG ESG Dataset Analysis Notebook
**Dataset:** Extracted from LSEG Workspace (TR.ESGScore, TR.Scope1EstTotal, etc.)
**Objective:** Analyze ESG metrics, missing values, correlations, and year-wise trends.

---

### 🔹 Notebook Contents:
1. **Load and Inspect Data**
2. **General Statistics**
3. **Missing Values Analysis**
4. **Correlation Matrices**
5. **Year-wise Analysis**
6. **ESG-Specific Visualizations**
"""

"""
## 2️⃣ General Statistics
Compute descriptive statistics for numeric ESG columns, grouped by year.
"""

# %%
# Identify ESG-related numeric columns
esg_numeric_cols = [
    col for col in lseg_df.columns
    if any(esg_term in col.lower() for esg_term in [
        "esg", "scope", "co2", "emission", "carbon", "energy", "water", "waste", "ghg",
        "revenue", "employees", "profit", "score", "target", "price", "tonne", "use"
    ])
]

# Also include any other numeric columns that might be relevant
all_numeric_cols = lseg_df.select_dtypes(include=[np.number]).columns.tolist()
numeric_cols = list(set(esg_numeric_cols + all_numeric_cols))

# Check if FiscalYear column exists
year_col = "FiscalYear" if "FiscalYear" in lseg_df.columns else None
if not year_col:
    for col in lseg_df.columns:
        if "year" in col.lower():
            year_col = col
            break

if year_col:
    print(f"📌 Year column detected: **{year_col}**")
    lseg_df[year_col] = pd.to_numeric(lseg_df[year_col], errors="coerce").astype("Int64")
else:
    print("⚠️ No year column detected. Analyzing full dataset without year grouping.")

# Display general statistics
if numeric_cols:
    print("\n📌 **Descriptive Statistics for Numeric Columns:**")
    display(lseg_df[numeric_cols].describe().T)

    if year_col:
        print(f"\n📌 **Statistics by Year ({year_col}):**")
        display(lseg_df.groupby(year_col)[numeric_cols].describe().T)
else:
    print("⚠️ No numeric ESG columns found. Check your column names.")

# %% [markdown]
"""
## 3️⃣ Missing Values Analysis
Identify and visualize missing values in the ESG dataset.
"""

# %%
# Calculate missing values
missing_values = lseg_df[esg_numeric_cols].isna().sum().sort_values(ascending=False)
missing_percent = (lseg_df[esg_numeric_cols].isna().mean() * 100).sort_values(ascending=False)
missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "% Missing": missing_percent
}).query("`Missing Values` > 0")

print("📌 **Missing Values in ESG Numeric Columns:**")
display(missing_df)

# Plot missing values
plt.figure(figsize=(12, 6))
sns.barplot(
    x=missing_percent.index,
    y=missing_percent.values,
    color="skyblue"
)
plt.xticks(rotation=90)
plt.title("Percentage of Missing Values in ESG Numeric Columns")
plt.ylabel("Missing %")
plt.xlabel("Columns")
plt.tight_layout()
plt.savefig(OUTPUT_PLOTS / "lseg_missing_values.png", dpi=300, bbox_inches="tight")
plt.show()

# %% [markdown]
"""
## 4️⃣ Correlation Matrices
Compute and visualize correlations between ESG metrics.
"""

# %%
if len(numeric_cols) > 1:
    # Compute correlation matrix
    corr_matrix = lseg_df[numeric_cols].corr()

    print("📌 **Correlation Matrix (Top ESG Metrics):**")
    # Show top 10x10 for readability
    top_esg_cols = [col for col in numeric_cols if any(term in col.lower() for term in ["esg", "scope", "co2", "emission", "score"])]
    if len(top_esg_cols) > 1:
        display(corr_matrix.loc[top_esg_cols, top_esg_cols].style.background_gradient(cmap="coolwarm"))

    # Plot heatmap
    plt.figure(figsize=(14, 12))
    sns.heatmap(
        corr_matrix,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0,
        square=True,
        linewidths=0.5,
        annot_kws={"size": 8}
    )
    plt.title("Correlation Matrix of ESG Metrics")
    plt.tight_layout()
    plt.savefig(OUTPUT_PLOTS / "lseg_correlation_matrix.png", dpi=300, bbox_inches="tight")
    plt.show()

    # Highlight strong correlations (|r| > 0.7)
    strong_corr = []
    for i in range(len(numeric_cols)):
        for j in range(i+1, len(numeric_cols)):
            corr = corr_matrix.iloc[i, j]
            if abs(corr) > 0.7:
                strong_corr.append((numeric_cols[i], numeric_cols[j], corr))

    if strong_corr:
        print("\n📌 **Strong Correlations (|r| > 0.7):**")
        for col1, col2, corr in strong_corr:
            print(f"- {col1} ↔ {col2}: {corr:.2f}")
else:
    print("⚠️ Not enough numeric columns for correlation analysis.")

# %% [markdown]
"""
## 5️⃣ Year-wise Analysis
Analyze ESG trends and distributions by year.
"""

# %%
if year_col and numeric_cols:
    # Numeric trends by year
    print(f"📌 **Mean ESG Metrics by Year ({year_col}):**")
    mean_by_year = lseg_df.groupby(year_col)[numeric_cols].mean().T
    display(mean_by_year.style.background_gradient(cmap="YlGnBu"))

    # Plot trends for top ESG metrics
    top_esg_metrics = [col for col in numeric_cols if any(term in col.lower() for term in ["esg", "scope", "co2", "emission", "score"])]
    if top_esg_metrics:
        plt.figure(figsize=(14, 8))
        for col in top_esg_metrics[:5]:  # Plot first 5 ESG metrics
            sns.lineplot(
                data=lseg_df,
                x=year_col,
                y=col,
                label=col,
                marker="o"
            )
        plt.title(f"ESG Metrics Trends by Year ({year_col})")
        plt.ylabel("Value")
        plt.xlabel(year_col)
        plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.savefig(OUTPUT_PLOTS / "lseg_esg_trends_by_year.png", dpi=300, bbox_inches="tight")
        plt.show()

    # Categorical distributions by year (e.g., PolicyEmissions)
    categorical_esg_cols = [
        col for col in lseg_df.columns
        if any(term in col.lower() for term in ["policy", "commitment", "target", "certified"])
        and lseg_df[col].dtype == "object"
    ]
    for col in categorical_esg_cols[:3]:  # Plot first 3 categorical ESG columns
        if lseg_df[col].nunique() <= 20:  # Avoid high-cardinality columns
            plt.figure(figsize=(12, 6))
            sns.countplot(
                data=lseg_df,
                x=year_col,
                hue=col,
                palette="viridis"
            )
            plt.title(f"Distribution of **{col}** by Year")
            plt.xticks(rotation=45)
            plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
            plt.tight_layout()
            plt.savefig(OUTPUT_PLOTS / f"lseg_{col}_by_year.png", dpi=300, bbox_inches="tight")
            plt.show()
else:
    print("⚠️ No year column or numeric ESG columns available for year-wise analysis.")

# %% [markdown]
"""
## 6️⃣ ESG-Specific Visualizations
Deep dive into key ESG metrics.
"""

# %%
# ESG Score Distribution
if "TR.ESGScore" in lseg_df.columns:
    plt.figure(figsize=(10, 6))
    sns.histplot(
        data=lseg_df,
        x="TR.ESGScore",
        kde=True,
        bins=30,
        color="green"
    )
    plt.title("Distribution of ESG Scores")
    plt.xlabel("ESG Score")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.savefig(OUTPUT_PLOTS / "lseg_esg_score_distribution.png", dpi=300, bbox_inches="tight")
    plt.show()

# Emissions by Scope
scope_cols = [col for col in lseg_df.columns if "Scope" in col and "Total" in col]
if scope_cols:
    plt.figure(figsize=(12, 6))
    scope_data = lseg_df[scope_cols].dropna()
    sns.boxplot(data=scope_data, palette="Set2")
    plt.title("Distribution of Emissions by Scope")
    plt.ylabel("Emissions (tonnes CO2e)")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(OUTPUT_PLOTS / "lseg_emissions_by_scope.png", dpi=300, bbox_inches="tight")
    plt.show()

# Energy Use vs. Revenue
if "TR.EnergyUseTotal" in lseg_df.columns and "TR.Revenue" in lseg_df.columns:
    plt.figure(figsize=(10, 6))
    sns.scatterplot(
        data=lseg_df,
        x="TR.Revenue",
        y="TR.EnergyUseTotal",
        alpha=0.5,
        hue=year_col if year_col else None
    )
    plt.title("Energy Use vs. Revenue")
    plt.xlabel("Revenue (USD)")
    plt.ylabel("Total Energy Use")
    plt.tight_layout()
    plt.savefig(OUTPUT_PLOTS / "lseg_energy_vs_revenue.png", dpi=300, bbox_inches="tight")
    plt.show()

# %% [markdown]
"""
## 7️⃣ Industry and Regional Analysis
Analyze ESG metrics by industry and country (if available).
"""

# %%
# Check for industry and country columns
industry_col = None
country_col = None

for col in lseg_df.columns:
    if "industry" in col.lower() and industry_col is None:
        industry_col = col
    if "country" in col.lower() and country_col is None:
        country_col = col

if industry_col:
    print(f"📌 Industry column detected: **{industry_col}**")
    # ESG Score by Industry
    if "TR.ESGScore" in lseg_df.columns:
        plt.figure(figsize=(14, 8))
        sns.boxplot(
            data=lseg_df,
            x=industry_col,
            y="TR.ESGScore",
            palette="viridis"
        )
        plt.title(f"ESG Score Distribution by {industry_col}")
        plt.xticks(rotation=90)
        plt.tight_layout()
        plt.savefig(OUTPUT_PLOTS / "lseg_esg_by_industry.png", dpi=300, bbox_inches="tight")
        plt.show()

if country_col:
    print(f"📌 Country column detected: **{country_col}**")
    # Emissions by Country
    if scope_cols:
        plt.figure(figsize=(14, 8))
        country_emissions = lseg_df.groupby(country_col)[scope_cols].mean().reset_index()
        for scope in scope_cols:
            sns.barplot(
                data=country_emissions,
                x=country_col,
                y=scope,
                alpha=0.7,
                label=scope
            )
        plt.title(f"Average Emissions by {country_col}")
        plt.xticks(rotation=90)
        plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
        plt.tight_layout()
        plt.savefig(OUTPUT_PLOTS / "lseg_emissions_by_country.png", dpi=300, bbox_inches="tight")
        plt.show()

# %% [markdown]
"""
## 🔚 Summary and Insights
- **General Statistics**: Descriptive stats for ESG metrics, grouped by year.
- **Missing Values**: Identified columns with missing data and visualized percentages.
- **Correlation Matrix**: Discovered relationships between ESG metrics (e.g., Scope 1 vs. Scope 2 emissions).
- **Year-wise Trends**: Analyzed how ESG metrics evolve over time.
- **ESG-Specific Visualizations**: Explored distributions of ESG scores, emissions, and energy use.
- **Industry/Regional Analysis**: Compared ESG performance across industries and countries.

**Next Steps:**
- Investigate **high missing value columns** (e.g., Scope 3 emissions).
- Explore **strong correlations** for potential insights (e.g., ESG Score vs. emissions).
- Compare **industry trends** to identify leaders/laggards in ESG performance.
"""
```
