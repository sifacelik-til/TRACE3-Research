# CDP Data Exploration

This notebook provides an overview of the CDP data structure and basic exploratory analysis.

**Key Objectives:**
- Understand the data directory structure
- Load sample data from different years
- Inspect data formats (Excel, Parquet)
- Get basic statistics about the datasets

## 1. Setup and Imports

```python
import sys
import os

# Add src to path
sys.path.insert(0, '../src')

from data_engine.data_loader import CDPDataLoader, get_data_summary
from utils import print_data_info, set_analysis_style, find_numeric_columns, find_categorical_columns

# Set visualization style
set_analysis_style()

print("Setup complete!")
```

## 2. Initialize Data Loader

```python
# Initialize the data loader
loader = CDPDataLoader()

# Get available years
years = loader.get_year_folders()
print(f"Available years: {years}")
print(f"Data spans from {min(years)} to {max(years)} ({len(years)} years)")
```

## 3. Explore Directory Structure

```python
# Check structure of the base directory
base_path = loader.base_data_path
print(f"Base data path: {base_path}\n")

# List top-level contents
contents = os.listdir(base_path)
print(f"Total items in base directory: {len(contents)}\n")

# Separate folders and files
folders = [item for item in contents if os.path.isdir(os.path.join(base_path, item))]
files = [item for item in contents if os.path.isfile(os.path.join(base_path, item))]

print(f"Folders ({len(folders)}):")
for folder in sorted(folders):
    print(f"  - {folder}")

print(f"\nFiles ({len(files)}):")
for file in sorted(files)[:5]:
    print(f"  - {file}")
if len(files) > 5:
    print(f"  ... and {len(files)-5} more files")
```

## 4. Load and Inspect 2024 Data

```python
# Load 2024 data (most recent)
print("Loading 2024 data...\n")
data_2024 = loader.load_year_data(2024, category="all")

print(f"\nLoaded {len(data_2024)} datasets from 2024")
print("\nDatasets loaded:")
for key in sorted(data_2024.keys()):
    if data_2024[key] is not None:
        print(f"  - {key}: {data_2024[key].shape}")
```

## 5. Detailed Data Inspection

```python
# Select first non-None dataset for inspection
sample_key = None
sample_df = None

for key, df in data_2024.items():
    if df is not None and not df.empty:
        sample_key = key
        sample_df = df
        break

if sample_df is not None:
    print(f"Inspecting: {sample_key}\n")
    print_data_info(sample_df, sample_key)
```

## 6. Data Type Analysis

```python
if sample_df is not None:
    numeric_cols = find_numeric_columns(sample_df)
    categorical_cols = find_categorical_columns(sample_df)
    
    print(f"Numeric columns ({len(numeric_cols)}):")
    for col in numeric_cols[:10]:
        print(f"  - {col}")
    if len(numeric_cols) > 10:
        print(f"  ... and {len(numeric_cols)-10} more")
    
    print(f"\nCategorical columns ({len(categorical_cols)}):")
    for col in categorical_cols[:10]:
        print(f"  - {col}")
    if len(categorical_cols) > 10:
        print(f"  ... and {len(categorical_cols)-10} more")
```

## 7. Summary Statistics

```python
if sample_df is not None:
    summary = get_data_summary(sample_df)
    print(f"Data Summary for {sample_key}:")
    print(f"  Shape: {summary['shape']}")
    print(f"  Memory Usage: {summary['memory_usage_mb']:.2f} MB")
    print(f"\nNumber of unique values per column:")
    unique_counts = sample_df.nunique().sort_values(ascending=False)
    for col, count in unique_counts.head(10).items():
        print(f"  - {col}: {count}")
```
