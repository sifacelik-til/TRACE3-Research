## 2. FactSet Data Loading

### 2.1 Initialize FactSet Loader and Explore Categories

```python
# Initialize FactSet loader
fs_loader = FactSetDataLoader()

# Get available categories
categories = fs_loader.get_categories()
print(f"Available FactSet categories: {categories}")
print(f"\nTotal categories: {len(categories)}")
```

### 2.2 List All Folders

```python
# Get all available folders
folders = fs_loader.list_folders()
print(f"Total FactSet folders: {len(folders)}\n")
print("Sample folders:")
for folder in folders[:15]:
    print(f"  - {folder}")
if len(folders) > 15:
    print(f"  ... and {len(folders)-15} more folders")
```

### 2.3 Load Specific Categories

```python
# Load symbol/entity data for linking
print("Loading FactSet symbol data...")
try:
    entity_data = fs_loader.load_category('sym_entity', encoding='utf-8')
    print(f"\nLoaded {len(entity_data)} entity datasets")
    for key in list(entity_data.keys())[:5]:
        if entity_data[key] is not None:
            print(f"  - {key}: {entity_data[key].shape}")
except Exception as e:
    print(f"Error loading entity data: {str(e)}")
```

### 2.4 Load Supply Chain Data

```python
# Load supply chain relationship data
print("Loading FactSet supply chain data...")
try:
    supply_chain_folder = 'ent_supply_chain_v1_full_3856'
    supply_chain_df = fs_loader.load_folder(supply_chain_folder, encoding='utf-8')
    
    if supply_chain_df is not None:
        print(f"\nSupply chain data shape: {supply_chain_df.shape}")
        print(f"Columns: {list(supply_chain_df.columns)}")
        print(f"\nFirst few rows:")
        print(supply_chain_df.head())
except Exception as e:
    print(f"Error loading supply chain data: {str(e)}")
```

## 3. Trucost Data Loading

### 3.1 Initialize Trucost Loader

```python
# Initialize Trucost loader
trucost_loader = TrucostDataLoader()

# Check available files
files = trucost_loader.get_available_files()
print("Available Trucost files:")
print(f"\nCSV files: {len(files['csv'])}")
for file in files['csv']:
    print(f"  - {file}")

print(f"\nStata files: {len(files['dta'])}")
for file in files['dta']:
    print(f"  - {file}")
```

### 3.2 Load GHG Data

```python
# Load GHG data
print("Loading Trucost GHG data...")
ghg_df = trucost_loader.load_ghg_data(file_type='csv')

if ghg_df is not None:
    print(f"\nGHG Data Summary:")
    print_data_info(ghg_df, "Trucost GHG")
```

### 3.3 Load Non-GHG Data

```python
# Load non-GHG data
print("Loading Trucost non-GHG data...")
nonghg_df = trucost_loader.load_nonghg_data(file_type='csv')

if nonghg_df is not None:
    print(f"\nNon-GHG Data Summary:")
    print_data_info(nonghg_df, "Trucost Non-GHG")
```
